```python
import time

# ==============================
# TIME CONSUMPTION BENCHMARK
# ==============================

REPEAT = 1000

# ------------------------------
# Dijkstra
# ------------------------------
start = time.perf_counter()

for _ in range(REPEAT):
    dijkstra(GRAPH_5, START)

end = time.perf_counter()

dijkstra_time = (end - start) / REPEAT

print("===== DIJKSTRA =====")
print(f"Average execution time: {dijkstra_time:.10f} seconds")


# ------------------------------
# Bellman-Ford
# ------------------------------
start = time.perf_counter()

for _ in range(REPEAT):
    bellman_ford(GRAPH_5, START)

end = time.perf_counter()

bellman_time = (end - start) / REPEAT

print("\n===== BELLMAN-FORD =====")
print(f"Average execution time: {bellman_time:.10f} seconds")


# ------------------------------
# Floyd-Warshall
# ------------------------------
start = time.perf_counter()

for _ in range(REPEAT):
    floyd_warshall(GRAPH_5)

end = time.perf_counter()

floyd_time = (end - start) / REPEAT

print("\n===== FLOYD-WARSHALL =====")
print(f"Average execution time: {floyd_time:.10f} seconds")
```
